```python
import numpy as np
import stim
from tqdm import tqdm
```

```python
def weight(x):
    return bin(x).count('1')
def parity(x):
    return bin(x).count('1') % 2
def measure(x, log):
    return parity(x & log)
```

```python
stabilizers = [
    0b11110000,
    0b00001111,
    0b11001100,
    0b10101010,
]
p = 0.1
sump, perror = 0, 0
weights = {}
idealweights = {0: 1, 1: 8, 2: 28, 3: 56, 4: 70, 5: 56, 6: 28, 7: 8, 8: 1}
for i in tqdm(range(0, 2**8)):
    thisweight = weight(i)
    if thisweight not in weights:
        weights[thisweight] = 0
    thisp = p ** thisweight * (1-p) ** (8-thisweight)
    sump += thisp
    for stab in stabilizers:
        if measure(i, stab) == 1:
            perror += thisp
            weights[thisweight] += 1
            break
print(round(sump, 8), round(perror, 8))
print(weights)
print(idealweights)
```
