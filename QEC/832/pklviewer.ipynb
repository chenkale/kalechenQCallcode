```python
import numpy as np
import pickle
```

```python
dict = pickle.load(open('sim/run0_double_rand0_n2_er2.pkl', 'rb'))
```

```python
outdir = 'test.bin'
with open(outdir, 'wb') as f:
    for row in dict:
        comp = np.uint64(int(row[0], 2))
        f.write(np.uint64(comp).tobytes())
        f.write(np.uint32(row[1]).tobytes())
```

```python
import random
row = random.choice(dict)
print(row[0], row[1])
comp = np.int64(int(row[0], 2))
print(comp)
```

```python
recovered = format(comp, '0{}b'.format(len(row[0])))
print("Recovered bitstring:", recovered)
```
