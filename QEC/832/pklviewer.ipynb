```python
import numpy as np
import pickle
```

```python
schedules = pickle.load(open('singlepromotionschedules.pkl', 'rb'))
print(len(schedules))
```

```python
gauges = []
for schedule in schedules:
    stab = np.array(schedule, dtype = np.bool)[0]
    is_duplicate = False
    for g in gauges:
        if np.array_equal(stab, g) or np.array_equal(stab ^ np.array([1, 1, 1, 1, 1, 1, 1, 1]), g):
            is_duplicate = True
            break
    if is_duplicate:
        continue
    gauges.append(stab)
print(len(gauges))
```

```python
logops = [np.array([1, 1, 1, 1, 1, 1, 1, 1], dtype = np.bool)]
for schedule in schedules:
    logops.append(np.array(schedule, dtype = np.bool)[1])
print(len(logops))
for ro in logops:
    print(ro)
with open('logops.pkl', 'wb') as f:
    pickle.dump(logops, f)
```
