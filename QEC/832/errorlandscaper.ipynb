```python
import numpy as np
import pickle
```

```python
# For single promotion, length 14 array
hotspot = np.array([7.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5,0.5, 0.5])
print(np.mean(hotspot))
hotspoterrors700 = [np.roll(hotspot, i) for i in range(700)]
with open('hotspoterrors700.pkl', 'wb') as f:
    pickle.dump(hotspoterrors700, f)
```

```python
# For double promotion, length 15 array
hotspot = np.array([8, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5, 0.5,0.5, 0.5, 0.5])
print(np.mean(hotspot))
hotspoterrors2800 = [np.roll(hotspot, i) for i in range(2800)]
with open('hotspoterrors2800.pkl', 'wb') as f:
    pickle.dump(hotspoterrors2800, f)
```

```python
# For single promotion, length 14 array
ranked = np.array([0.35, 0.45, 0.55, 0.65, 0.75, 0.85, 0.95, 1.05, 1.15, 1.25, 1.35, 1.45, 1.55, 1.65])
print(np.mean(ranked))
rankederrors700 = [np.random.permutation(ranked) for i in range(700)]
with open('rankederrors700.pkl', 'wb') as f:
    pickle.dump(rankederrors700, f)
```
