```python
# encode stabilizers

xxxx = np.array([[1,1,1,1], [0,0,0,0]])
zzzz = np.array([[0,0,0,0], [1,1,1,1]])
xixi = np.array([[1,0,1,0], [0,0,0,0]])
ziii = np.array([[0,0,0,0], [1,0,0,0]])
stabilizers = [xxxx, zzzz]

def commutes(a, b):
    sum = 0
    sum += np.sum(a[0] * b[1])
    sum += np.sum(a[1] * b[0])
    return sum % 2 == 0
def commutes_all(a, oplist):
    if type(oplist[0]) == tuple or isinstance(oplist[0], (list, np.ndarray)) and len(oplist[0]) > 0 and isinstance(oplist[0][0], (np.ndarray, list)):
        for row in oplist:
            for op in row:
                if not commutes(a, op):
                    return False
    else:
        for op in oplist:
            if not commutes(a, op):
                return False
    return True
def distinct(a, b):
    for op in stabilizers:
        if np.array_equal((a + op) % 2, b):
            return False
    return True

def equals_stabilizer(a):
    for op in stabilizers:
        if np.array_equal(a, op):
            return True
    return False
def printop(operator):
    print(str(operator[0])[1:-1] + ' | ' + str(operator[1])[1:-1])
def printops(operators):
    if type(operators[0]) == tuple:
        for tup in operators:
            for op in tup:
                printop(op)
            print()
    else:
        for op in operators:
            printop(op)

print(commutes(xxxx, ziii))

```
