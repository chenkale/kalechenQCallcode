```python
def hamming7(noise, nrounds):
    cb = CircuitBuilder.CircuitBuilder(noise, noise, noise, noise, noise, noise)

    # State preparation
    cb.al('R 0 1 2 3 4 5 6')
    cb.al('R 7 8 9 10 11 12')
    cb.al('H 10 11 12')

    cb.begin_loop(nrounds)  
    
    cb.ale('CX 0 7 1 7 2 7 3 7') # Stabilizer 1
    cb.ale('CX 0 8 1 8 4 8 5 8') # Stabilizer 2
    
    cb.ale('CX 0 9 2 9 4 9 6 9') # Stabilizer 3
    
    cb.ale('CX 10 0 10 1') # Half of stabilizer 4
    cb.ale('CX 10 2 10 3') # Half of stabilizer 4
    #cb.ale('CX 10 0 10 1 10 2 10 3') # Stabilizer 4
    cb.ale('CX 11 0 11 1 11 4 11 5') # Stabilizer 5
    cb.ale('CX 12 0 12 2 12 4 12 6') # Stabilizer 6

    cb.ale(f'MR({noise}) 7 8 9')
    cb.ale('H 10 11 12')
    cb.ale(f'MR({noise}) 10 11 12')
    cb.ale('H 10 11 12')

    cb.end_loop()
    return cb
```
