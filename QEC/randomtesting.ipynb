```python
# Model ZZZZ stabilizer check
c = cb.CircuitBuilder(0.01, 0.01, 0.01, 0.01, 0.01, 0.01)

#c.ale('R 0 1 2 3')
c.ale('CX 0 1')
c.ale('CX 0 2')
c.ale('CX 0 3')
c.al('M 0 1 2 3')
c.add_detector([-1, -2, -3, -4])
print(c.sample(10000))
sample = c.get_circuit().compile_detector_sampler().sample(10000)
print(np.sum(sample[:,0]))
# Model
```
